```python
trades = pd.read_csv("historical_data.csv")
sentiment = pd.read_csv("fear_greed_index.csv")
```

```python
trades['Timestamp'] = pd.to_datetime(trades['Timestamp'], unit='ms', errors='coerce')
trades['date'] = trades['Timestamp'].dt.date
sentiment['date'] = pd.to_datetime(sentiment['date'], errors='coerce').dt.date
```

```python
df = trades.merge(
    sentiment,
    on='date',
    how='left'
)
```

```python
print(f"Trades date range: {trades['date'].min()} to {trades['date'].max()}")
print(f"Sentiment date range: {sentiment['date'].min()} to {sentiment['date'].max()}")
```

```python
df['classification'].isnull().sum()
```

```python
df['win'] = df['Closed PnL'] > 0
```

```python
df['volume'] = (
    df['Execution Price'] *
    df['Size USD']
)
```

```python
print('Mean Closed PnL by Classification:')
print(df.groupby('classification')['Closed PnL'].mean())
```

```python
print('\nWin Rate by Classification (%):')
print(df.groupby('classification')['win'].mean() * 100)
```
